##Outlier Capping
This cell reduces the effect of extreme vaues by applying outlier capping and remove extra values which are not required.


```python
def outlier_capping(df,column):
    Q1= df[column].quantile(0.25)
    Q3= df[column].quantile(0.75)
    IQR= Q3-Q1
    lower_extreme= Q1-1.5*IQR
    upper_extreme= Q3+1.5*IQR
    df[column]= df[column].apply(lambda x: lower_extreme if x< lower_extreme else upper_extreme if x>upper_extreme else x)
for col in features.select_dtypes(include=['int','float']).columns:
    outlier_capping(features,col)
```

## Prepare numerical features

- `Country` is a name, so it is excluded from numeric clustering features.
- Currency-like fields such as GDP and tourism values contain `$` and commas. Convert them to numeric values.
- `Number of Records` is a record counter rather than a development measurement, so exclude it from clustering.
- Missing numeric values are filled with the median.

```python
data = df.copy()

# Convert currency-formatted columns to numeric values
currency_columns = [
    "GDP", "Health Exp/Capita", "Tourism Inbound", "Tourism Outbound"
]

for col in currency_columns:
    if col in data.columns:
        data[col] = (
            data[col].astype(str)
            .str.replace("$", "", regex=False)
            .str.replace(",", "", regex=False)
            .replace({"nan": np.nan, "None": np.nan, "": np.nan})
        )
        data[col] = pd.to_numeric(data[col], errors="coerce")

# Keep numeric columns only; exclude identifiers/counters
exclude_columns = ["Country", "Number of Records"]
numeric_data = data.drop(columns=exclude_columns, errors="ignore")
numeric_data = numeric_data.select_dtypes(include=np.number)

# Remove columns that are entirely missing or constant
numeric_data = numeric_data.dropna(axis=1, how="all")
constant_columns = [
    col for col in numeric_data.columns
    if numeric_data[col].nunique(dropna=True) <= 1
]
numeric_data = numeric_data.drop(columns=constant_columns)

print("Features used:", numeric_data.columns.tolist())
print("Number of features:", numeric_data.shape[1])
print("Removed constant columns:", constant_columns)
display(numeric_data.head())
```

##Impute missing values and scale features

Scaling is important because the features use different units and ranges. Median imputation fills missing values, then `StandardScaler` places numeric features on comparable scales.

```python
imputer = SimpleImputer(strategy="median")
X_imputed = imputer.fit_transform(numeric_data)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_imputed)

print("Scaled data shape:", X_scaled.shape)
print("Missing values remaining:", np.isnan(X_scaled).sum())
```

## Applying PCA before clustering

PCA reduces the feature dimensions while retaining as much variance as possible. We retain 95% of the variance for the clustering input.

The separate two-component PCA plot used later is only for visualization.

```python
pca = PCA(n_components=0.95, random_state=42)
X_pca = pca.fit_transform(X_scaled)

print("Original number of features:", X_scaled.shape[1])
print("PCA components retained:", X_pca.shape[1])
print("Total explained variance:", round(pca.explained_variance_ratio_.sum(), 4))
```

## Method 2: Agglomerative Clustering

Agglomerative clustering builds clusters by repeatedly merging similar groups. Ward linkage attempts to minimize the increase in within-cluster variance.

```python
agg_model = AgglomerativeClustering(n_clusters=3, linkage="ward")
agg_labels = agg_model.fit_predict(X_pca)

print("Agglomerative cluster counts:")
display(pd.Series(agg_labels, name="Cluster").value_counts().sort_index())
```

## Method 3: DBSCAN

DBSCAN finds dense regions and can mark outliers as noise (`-1`). It does not require the number of clusters in advance. `eps=0.5` and `min_samples=5` are starting parameters and may need tuning for this dataset.

```python
dbscan = DBSCAN(eps=0.5, min_samples=5)
dbscan_labels = dbscan.fit_predict(X_pca)

print("DBSCAN labels (including -1 for noise):")
display(pd.Series(dbscan_labels, name="Cluster").value_counts().sort_index())
print("Noise points:", int(np.sum(dbscan_labels == -1)))
```

```python
agg_model = AgglomerativeClustering(n_clusters=3, linkage="ward")
agg_labels = agg_model.fit_predict(X_pca)

print("Agglomerative cluster counts:")
display(pd.Series(agg_labels, name="Cluster").value_counts().sort_index())
```

## Method 4: Gaussian Mixture Model (GMM)

GMM models the observations as a mixture of probability distributions. Each observation is assigned to the component with the highest estimated probability.

```python
# Create GMM model
gmm = GaussianMixture(n_components=3, random_state=42)

# Train the model and predict clusters
gmm_labels = gmm.fit_predict(X_pca)

# Evaluate the model
silhouette = silhouette_score(X_pca, gmm_labels)
db_index = davies_bouldin_score(X_pca, gmm_labels)
ch_score = calinski_harabasz_score(X_pca, gmm_labels)
```

## Step 11 — Evaluate and compare the four methods

The metrics below are computed on the PCA-transformed data used to train the clustering models. For DBSCAN, noise points are excluded from the metric calculation; its noise count is reported separately. A score is left as `NaN` when the method produces fewer than two valid clusters or otherwise cannot be evaluated.

```python
models = {
    "K-Means": kmeans_labels,
    "Agglomerative": agg_labels,
    "DBSCAN": dbscan_labels,
    "GMM": gmm_labels
}

results = []

for method_name, labels in models.items():
    labels = np.asarray(labels)

    if method_name == "DBSCAN":
        mask = labels != -1
        eval_X = X_pca[mask]
        eval_labels = labels[mask]
        noise_points = int(np.sum(labels == -1))
    else:
        eval_X = X_pca
        eval_labels = labels
        noise_points = 0

    unique_clusters = np.unique(eval_labels)
    n_clusters = len(unique_clusters)

    # Silhouette and the other metrics require at least 2 clusters,
    # and not one cluster per observation.
    if 2 <= n_clusters < len(eval_labels) and len(eval_labels) > 1:
        sil = silhouette_score(eval_X, eval_labels)
        dbi = davies_bouldin_score(eval_X, eval_labels)
        ch = calinski_harabasz_score(eval_X, eval_labels)
    else:
        sil, dbi, ch = np.nan, np.nan, np.nan

    results.append({
        "Method": method_name,
        "Number of Clusters (excluding noise)": n_clusters,
        "Noise Points": noise_points,
        "Silhouette Score (higher is better)": sil,
        "Davies-Bouldin Index (lower is better)": dbi,
        "Calinski-Harabasz Score (higher is better)": ch
    })

comparison_df = pd.DataFrame(results)
display(comparison_df.round(4))
```

## Select the provisional best method

We use the highest valid Silhouette Score as the primary ranking rule. Review the other metrics, the number of clusters, DBSCAN's noise count, and the visualizations before finalizing the model.

```python
valid_comparison = comparison_df.dropna(
    subset=["Silhouette Score (higher is better)"]
)

if len(valid_comparison) == 0:
    print("No method has a valid Silhouette Score. Review model parameters.")
else:
    best_row = valid_comparison.loc[
        valid_comparison["Silhouette Score (higher is better)"].idxmax()
    ]
    print("Provisional best method:", best_row["Method"])
    print("Silhouette Score:", round(best_row["Silhouette Score (higher is better)"], 4))
    print("Clusters:", int(best_row["Number of Clusters (excluding noise)"]))
    print("Noise points:", int(best_row["Noise Points"]))
```

##Using Elbow Method of Choose K Value


```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia_values = []

k_values = range(2, 11)

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X_pca)

    inertia_values.append(model.inertia_)

plt.figure(figsize=(8, 5))

plt.plot(
    list(k_values),
    inertia_values,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")

plt.title("Elbow Method for Optimal K")

plt.xticks(list(k_values))

plt.grid(True)

plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

k_values = range(2, 11)

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_pca)

    score = silhouette_score(
        X_pca,
        labels
    )

    silhouette_scores.append(score)

    print(
        f"K = {k}, Silhouette Score = {score:.4f}"
    )

plt.figure(figsize=(8, 5))

plt.plot(
    list(k_values),
    silhouette_scores,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")

plt.title("Silhouette Score for Different K Values")

plt.xticks(list(k_values))

plt.grid(True)

plt.show()
```

```python
%%writefile world_development_deployment/app.py
# Create a Streamlit application that applies the same preprocessing pipeline.
#App_Code

import streamlit as st
import pandas as pd
import numpy as np
import joblib
from pathlib import Path
import matplotlib.pyplot as plt

# ----------------------------------
# 1. Page Configuration
# ----------------------------------

st.set_page_config(
    page_title="P712 World Development Clustering",
    page_icon="🌍",
    layout="wide"
)

BASE_DIR = Path(__file__).resolve().parent

# ----------------------------------
# 2. Load Saved Models
# ----------------------------------

@st.cache_resource
def load_models():
    imputer = joblib.load(BASE_DIR / "imputer.pkl")
    scaler = joblib.load(BASE_DIR / "scaler.pkl")
    pca = joblib.load(BASE_DIR / "pca.pkl")
    model = joblib.load(BASE_DIR / "kmeans_model.pkl")
    features = joblib.load(BASE_DIR / "feature_names.pkl")

    return imputer, scaler, pca, model, features

try:
    imputer, scaler, pca, model, features = load_models()
except Exception as e:
    st.error(f"Unable to load model files: {e}")
    st.stop()

# ----------------------------------
# 3. App Header
# ----------------------------------

st.title("🌍 World Development Clustering")
st.subheader("P712 - Unsupervised Machine Learning Project")

st.write(
    "Explore development indicators and predict "
    "the cluster assigned by the trained K-Means model."
)

st.info(
    "Enter values using the same units and feature definitions "
    "as the dataset used to train the model."
)

# ----------------------------------
# 4. Sidebar
# ----------------------------------

st.sidebar.header("Project Information")
st.sidebar.write("**Project:** P712 Clustering")
st.sidebar.write("**Algorithm:** K-Means")
st.sidebar.write("**Learning Type:** Unsupervised Learning")
st.sidebar.write(f"**Features:** {len(features)}")
st.sidebar.write(f"**Clusters:** {model.n_clusters}")

page = st.sidebar.radio(
    "Choose Page",
    ["Cluster Prediction", "Model Information"]
)

# ----------------------------------
# 5. Cluster Prediction Page
# ----------------------------------

if page == "Cluster Prediction":

    st.header("🔍 Predict Development Cluster")

    st.write(
        "Enter the numerical values for the development indicators."
    )

    with st.form("prediction_form"):

        user_values = {}

        col1, col2 = st.columns(2)

        for i, feature in enumerate(features):

            with col1 if i % 2 == 0 else col2:

                user_values[feature] = st.number_input(
                    label=str(feature),
                    value=0.0,
                    format="%.4f",
                    key=f"input_{i}"
                )

        submitted = st.form_submit_button(
            "Predict Cluster",
            type="primary"
        )

    if submitted:

        input_df = pd.DataFrame(
            [[user_values[f] for f in features]],
            columns=features
        )

        try:
            # Use the same preprocessing order as training
            imputed = imputer.transform(input_df)
            scaled = scaler.transform(imputed)
            reduced = pca.transform(scaled)

            prediction = model.predict(reduced)

            cluster = int(prediction[0])

            st.success(f"Predicted Cluster: {cluster}")

            st.subheader("Input Summary")
            st.dataframe(input_df, use_container_width=True)

            st.caption(
                "Cluster numbers are identifiers, not development rankings."
            )

        except Exception as e:
            st.error(f"Prediction failed: {e}")

# ----------------------------------
# 6. Model Information Page
# ----------------------------------

elif page == "Model Information":

    st.header("📊 Model Information")

    col1, col2, col3 = st.columns(3)

    col1.metric("Algorithm", "K-Means")
    col2.metric("Number of Clusters", model.n_clusters)
    col3.metric("Input Features", len(features))

    st.subheader("Features Used")

    st.write(features)

    st.subheader("Preprocessing Pipeline")

    st.markdown(
        """
        1. Missing value handling using the saved imputer
        2. Feature scaling using the saved StandardScaler
        3. Dimensionality reduction using the saved PCA
        4. Cluster prediction using the saved K-Means model
        """
    )

    st.subheader("Cluster Centroids")

    centroids = pd.DataFrame(
        model.cluster_centers_,
        columns=[
            f"PC{i + 1}"
            for i in range(model.cluster_centers_.shape[1])
        ]
    )

    st.dataframe(centroids, use_container_width=True)

    fig, ax = plt.subplots(figsize=(9, 4))

    for i, row in centroids.iterrows():
        ax.plot(
            row.index,
            row.values,
            marker="o",
            label=f"Cluster {i}"
        )

    ax.set_title("K-Means Centroids in PCA Space")
    ax.set_xlabel("Principal Components")
    ax.set_ylabel("Centroid Value")
    ax.legend()
    ax.grid(True, alpha=0.3)

    st.pyplot(fig)
    plt.close(fig)

# ----------------------------------
# 7. Footer
# ----------------------------------

st.divider()

st.caption(
    "P712 - World Development Clustering | "
    "Built with Python, Pandas, Scikit-learn and Streamlit"
)
```
